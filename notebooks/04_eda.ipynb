# 04 - Exploratory Data Analysis

Explore the **training** split only, to avoid leaking information from
the test set into modelling decisions.

```python
import os
from pathlib import Path

# Mirror the same path convention used in the other notebooks
path = os.getcwd()
path = os.path.abspath(os.path.join(path, "..", "data"))

DATA_DIR      = Path(path)
PROCESSED_DIR = DATA_DIR / "processed"

TRAIN_FILE    = PROCESSED_DIR / "train.parquet"
TEST_FILE     = PROCESSED_DIR / "test.parquet"

TARGET_COLUMN = "label"
```

```python
import pandas as pd
import matplotlib.pyplot as plt

train = pd.read_parquet(TRAIN_FILE)
print(f"Train shape: {train.shape}")
train.head()
```

## Summary statistics

```python
train.describe(include="all").T
```

## Target balance

```python
target = TARGET_COLUMN
if target in train.columns:
    train[target].value_counts().plot(kind="bar")
    plt.title(f"Distribution of {target}")
    plt.show()
```

## Numeric feature distributions

```python
num_cols = train.select_dtypes(include="number").columns
train[num_cols].hist(figsize=(12, 8), bins=30)
plt.tight_layout()
plt.show()
```

## Correlations

```python
corr = train.select_dtypes(include="number").corr()
fig, ax = plt.subplots(figsize=(8, 6))
im = ax.imshow(corr, cmap="coolwarm", vmin=-1, vmax=1)
ax.set_xticks(range(len(corr)))
ax.set_xticklabels(corr.columns, rotation=90)
ax.set_yticks(range(len(corr)))
ax.set_yticklabels(corr.columns)
fig.colorbar(im)
plt.title("Feature correlation")
plt.tight_layout()
plt.show()
```
